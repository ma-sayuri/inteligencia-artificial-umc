**20. Matriz Transposta**

```python
matriz = [[1, 2, 3], [4, 5, 6]]
# cria lista vazia
transposta = []

for i in range(len(matriz[0])):
    linha = []
    for j in range(len(matriz)):
        linha.append(matriz[j][i])

    transposta.append(linha)

print(f"Matriz original: {matriz}")
print(f'Matriz Transposta: {transposta}')
```

**21. Busca Binária**

```python
lista_num = [5, 10, 15, 20, 25]
num_alvo = 10
print(f'lista de numeros: {lista_num}\nnumero alvo: {num_alvo}')
inicio = 0
fim = len(lista_num) - 1
encontrado = False

# continua procurando enquanto tiver elemento para verificar
while inicio <= fim:
    meio = (inicio + fim) // 2   # encontra a posicao do meio da lista

    if lista_num[meio] == num_alvo:
        print("Elemento encontrado na posicao:", meio)
        encontrado = True     # elemento foi encontrado
        break
    elif lista_num[meio] < num_alvo:
        inicio = meio + 1
    else:
        fim = meio - 1    # procura na parte esquerda da lista

# verifica se o elemento nao foi encontrado
if not encontrado:
    print("Elemento nao encontrado")
```

> *=> Primeiro, define uma lista de numeros e um número alvo, inicializando os índices de início e fim. Depois, em um laço while a lista é dividida ao meio e o valor central é comparado ao alvo (a cada repetição). Se for igual, sua posição é exibida, se for menor ou maior, a busca continua na metade direita ou esquerda. No final, exibe se não foi encontraso*

**22. Criptografia de Cifra de César**

```python
msg = input('Digite a mensagem a ser criptografada: ')
chave = int(input('Digite a chave de deslocamento: '))
msg_cifrada = ""

# passa por cada caractere da mensagem
for caractere in msg:
    if caractere.isalpha():   # verifica se e uma letra
        # define ponto de partida do alfabeto
        base = ord('A') if caractere.isupper() else ord('a')
        # calcula o deslocamento dentro das 26 letras do alfabeto
        novo_char = chr((ord(caractere) - base + chave) % 26 + base)
        msg_cifrada += novo_char
    else:
        # mantem espacos e pontuacoes sem alterar
        msg_cifrada += caractere

print(f'Mensagem criptografada: {msg_cifrada}')
```

> *=> Solicita a mensagem e o valor inteiro da chave de deslocamento ao usuário. Depois, percorre o texto caractere por caractere com um laço for. Para cada letra, usa a função ord() para pegar o código numérico ASCII e identifica se é maiúscula ou minúscula com .isupper(). Aplica a fórmula da Cifra de César usando % 26 (se passar de Z ele volta ao início do alfabeto) e converte o número para caractere com chr(). No final, exibe a mensagem criptografada*

**23. Descriptografar Cifra de César**

```python
msg_cifrada = input('Digite a mensagem cifrada: ')
chave = int(input('Digite a chave de deslocamento utilizada antes: '))
msg_original = ""

# passa por cada caractere da mensagem cifrada
for caractere in msg_cifrada:
    if caractere.isalpha():
        base = ord('A') if caractere.isupper() else ord('a')
        # subtrai a chave para desfazer o deslocamento
        novo_char = chr((ord(caractere) - base - chave) % 26 + base)
        msg_original += novo_char
    else:
        msg_original += caractere

print(f'Mensagem descriptografada: {msg_original}')
```

> *=> Primerio solicita a mensagem criptografada e a mesma chave usada antees ao usuário. Depois, percorre cada caractere, e em vez de somar a chave como no exercício 22, subtrai a chave do código da letra. No final do laço, exibe a mensagem original revelada*

**24. Validação de CPF**

```python
def validar_cpf(cpf):
  # caso o tamanho do cpf seja invalido
  if(len(cpf)!=11):
    return False
  # caso nao contenha apenas numeros
  if(not cpf.isnumeric()):
    return False

  digitos_cpf = cpf[:9]
  digitos_verificadores = cpf[9:]
  digitos_verificadores_obtidos = ""
  soma = 0
  restoSoma = 0

  # soma o produto dos nove primeiros digitos pela sequencia de 10 a 2
  for i in range(10,1,-1):
    soma += int(digitos_cpf[10-i])*i
  # define o primeiro digito verificador
  restoSoma = soma%11
  if(restoSoma < 2):
    digitos_verificadores_obtidos += "0"
  else:
    digitos_verificadores_obtidos += f"{11-restoSoma}"

  # soma o produto dos nove primeiros digitos + primeiro digito verificador pela sequencia de 11 a 2
  digitos_cpf += digitos_verificadores_obtidos
  soma = 0
  for i in range(11,1,-1):
    soma += int(digitos_cpf[11-i])*i

  # define o segundo digito verificador
  restoSoma = soma%11
  if(restoSoma<2):
    digitos_verificadores_obtidos += "0"
  else:
    digitos_verificadores_obtidos += f"{11-restoSoma}"

  # retorna a verificacao
  return digitos_verificadores_obtidos==digitos_verificadores

cpf = input('Informe seu cpf (apenas numeros, sem pontuacao): ')
print(f'O cpf informado e {"valido" if validar_cpf(cpf) else "invalido"}')
```

> *=> Cria uma função que primeiro verifica se o cpf possui 11 dígitos e contém apenas números. Depois, realiza os cálculos do algoritmo oficial, multiplicando os 9 primeiros dígitos por uma sequência decrescente, e calculando o % por 11 para definir os dois dígitos verificadores. No final, compara os dígitos com os que foram informados para validar*

**25. Classificação de Números**

```python
lista_num = [-5, -4, -3, -2, -1, 0, 0,0, 1, 2, 3, 4, 5]
classificacao = {
    "positivos":[],
    "negativos":[],
    "zeros":0
}

for n in lista_num:
  if(n==0):
    classificacao["zeros"]+=1
  elif(n>0):
    classificacao["positivos"].append(n)
  else:
    classificacao["negativos"].append(n)

print(lista_num)
print(f"Numeros positivos: {classificacao["positivos"]}")
print(f"Numeros negativos: {classificacao["negativos"]}")
print(f"Quantidade de zeros na lista: {classificacao["zeros"]}")
```

```python
palavra = "pokemon"
letras_certinhas = ["_"] * len(palavra)
tentativas = 6
letras_digitadas = []
print("jogo da forca")

# so roda enquanto restarem tentativas e se ainda tiver letra escondida
while tentativas > 0 and "_" in letras_certinhas:
    print("\nPalavra:", " ".join(letras_certinhas))
    print(f"tentativas restantes: {tentativas}")
    letra = input("\nDigite uma letra: ").lower()

    if letra in letras_digitadas:
        print("opa, essa letra ja foi")
        continue

    letras_digitadas.append(letra)

    # verifica se a letra informada esta na palavra
    if letra in palavra:
        for i in range(len(palavra)):
            if palavra[i] == letra:
                letras_certinhas[i] = letra
        print("a letra esta na palavra")
    else:
        tentativas -= 1
        print("a letra nao esta na palavra")

# exibe se acertou ou perdeu
if "_" not in letras_certinhas:
    print(f"\nACERTOU, a palavra era: {palavra}")
else:
    print(f"\nACABOU... a palavra era: {palavra}")
```

**31. Ordenação por Bolha**

```python
lista_num = [5, 3, 4, 6, 2, 1]
print(f'lista: {lista_num}')

# compara os números vizinhos e troca eles de posição
for i in range(len(lista_num)):
    for j in range(len(lista_num) - 1 - i):
        if lista_num[j] > lista_num[j + 1]:
            temp = lista_num[j]
            lista_num[j] = lista_num[j + 1]
            lista_num[j + 1] = temp

print(f'lista ordenada: {lista_num}')
```

**32. Jogo da Velha**

```python
tabuleiro = [" "]*9

def mostrar_tabuleiro():
   print(f"{tabuleiro[0]}|{tabuleiro[1]}|{tabuleiro[2]}")
   print(f"{tabuleiro[3]}|{tabuleiro[4]}|{tabuleiro[5]}")
   print(f"{tabuleiro[6]}|{tabuleiro[7]}|{tabuleiro[8]}")

jogador = "X"
jogo_terminou = False
# combinacoes que formam uma vitoria
vitoria = [(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]
rodadas = 0

while not jogo_terminou and rodadas < 9:
   mostrar_tabuleiro()
   posicao = int(input(f"\njogador {jogador}, escolha uma posicao de 0 a 8): "))

   if tabuleiro[posicao] != " ":
       print("posicao ja ocupada... tente de novo")
       continue
   tabuleiro[posicao] = jogador
   rodadas += 1

   for combinacao in vitoria:
       a, b, c = combinacao
       if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
           mostrar_tabuleiro()
           print(f"\nJogador {jogador} venceu!!")
           jogo_terminou = True
           break
   jogador = "O" if jogador == "X" else "X" # alterna o turno

if not jogo_terminou:
   mostrar_tabuleiro()
   print("Deu velha!!")
```

> *=> Primeiro inicializa uma lista de 9 posições vazias para simular o tabuleiri e define as combinações de vitória. O laço while principal roda até 9 rodadas ou até alguém vencer*

**37. Funções Recursivas**

```python
def fatorial_recursivo(n):
    # fatorial de 0 ou 1
    if n == 0 or n == 1:
        return 1
    # o numero * o fatorial do seu antecessor
    else:
        return n * fatorial_recursivo(n - 1)

# solicita o numero
numero = int(input('Digite um numero inteiro: '))
# chama a funcao e exibee
resultado = fatorial_recursivo(numero)
print(f'O fatorial de {numero} é: {resultado}')
```

> *=> Define a função fatorial_recursivo, que verifica primeiro se o número for 0 ou 1 (retorna 1), para evitar um loop infinito. 48. Ordenação por Inserção

```python
lista = [7, 5, 9, 10, 8, 6]

# percorre inserindo cada numero na posicao correta da lista
for i in range(1, len(lista)):
    chave = lista[i]
    j = i - 1

    while j >= 0 and lista[j] > chave:
        lista[j + 1] = lista[j]
        j -= 1

    lista[j + 1] = chave

print("Lista ordenada:", lista)
```

> *=> Primeiro, cria a lista de números e o laço for começa do segundo elemento (índice 1), definindo como a 'chave'. O laço while dentro verifica os elementos a esquerda da chave, movendo os maiores uma posição para a direita. Quando encontrar a posição correta, a chave é inserida. O processo manda os números menores para o início*

**50. Cálculo de Frequência de Letras**

```python
# pede uma string
texto = input('Digite um texto para saber a frequencia de cada letra nele: ')
count_freq = {}

# percorre o texto e salva no dicionario a contagem
for l in texto:
  # isalpha para pegar so letra e desconsiderar os espacos
  if l.isalpha():
    if l in count_freq:
      count_freq[l] += 1
    else:
      count_freq[l] = 1

for key, value in count_freq.items():
    print(f"{key}: {value}")
```
