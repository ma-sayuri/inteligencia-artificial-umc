**19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado)**

```python
import numpy as np

# dois arrays de tamanho 20 com numeros ints aleatorios
arrayA = np.random.randint(1, 20, 20)
arrayB = np.random.randint(1, 20, 20)
print(f'arrayA de tamanho 20 com numeros aleatorios:\n{arrayA}\narrayB de tamanho 20 com numeros aleatorios:\n{arrayB}')

# calcula produto escalar
produto_escalar = np.dot(arrayA, arrayB)
# norma dos vetores
norma_arrayA = np.linalg.norm(arrayA)
norma_arrayB = np.linalg.norm(arrayB)
# cosseno do angulo
cos_t = produto_escalar / (norma_arrayA*norma_arrayB)

# encontra angulo em radianos
angulo_radi = np.arccos(cos_t)
angulo_graus = np.degrees(angulo_radi)  # converte para graus
print(f'\nangulo em radianos: {angulo_radi:.4f}\nangulo em graus: {angulo_graus:.2f}°')
```

> *=> Primeiro cria dois arrays de 20 números aleatórios. Depois, calcula o produto escalar com np.dot() e a norma com np.linalg.norm(). Usa esses valores para encontrar o cosseno do ângulo. No final, extrai o ângulo em radianos com np.arccos() e converte para graus com np.degrees(), e exibe ambos*

**1. Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π**

```python
import numpy as np
from scipy import integrate

# define a funcao
def f(x):
    return (x**2) * np.sin(x)

# calcula integral de 0 a pi
result, erro = integrate.quad(f, 0, np.pi)
print('resultado da integral:', result)
print('erro estimado:', erro)
```

> *=> Primeiro, cria a função matemática dada. Depois, para calcular a integral definida, passa ela para a função integrate.quad() junto com os limites de 0 e np.pi. No final, exibe o valor calculado e o erro estimado*

**2. Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0 a 10**

```python
import numpy as np
from scipy.integrate import solve_ivp

# define a equacao diferencial
def eq_diferencial(x, y):
    return x - y

# resolve no intervalo 0, 10 com y(0) = 1
solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

# exibe 10 primeiros resultados
print('solucao para y:', solucao.y[0][:10])
```

> *=> Define a equação diferencial na função eq_diferencial. Utiliza a função solve_ivp(), onde passa a equação, o intervalo [0, 10], a condição onde y é igual a [1]. No final, exibe só os 10 primeiros resultados*

**3. Encontre as raízes da função f(x) = x⁴ − 3x³ + 2 utilizando SciPy**

```python
from scipy.optimize import root

# define a funcao
def f(x):
    return x**4 - 3*(x**3) + 2

# encontra a raiz proxima de x = 0
solucao = root(f, 0)
print('raiz encontrada:', solucao.x)
```

**4. Use a função de otimização de SciPy para maximizar f(x) = −x²+ 4x + 1**

```python
from scipy.optimize import minimize

# define o negativo da funcao para usar o minimize
def f_invertida(x):
    return -(-x**2 + 4*x + 1)

# usa metodo de minimizacao
resultado = minimize(f_invertida, 0)
print('ponto de maximo encontrado em x =', resultado.x)
```

**7. Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0), (1, 2), (2, 3), (3, 5), (4, 4)**

**9. Ajuste uma curva quadrática aos pontos (1, 2), (2, 4), (3, 5), (4, 7), (5, 8)**

**10. Calcule a integral dupla da função f(x, y) = xy + x² no intervalo [0, 2] para x e [0, 1] para y**

```python
from scipy import integrate

# define a funcao
def f(y, x):
    return x * y + x**2

# calcula a integral dupla
result, erro = integrate.dblquad(f, 0, 2, lambda x: 0, lambda x: 1)

print('resultado da integral dupla:', result)
print('erro estimado:', erro)
```

**14. Utilize a função de otimização multivariada para minimizar f(x, y) = (x − 2)² + (y − 3)²**

```python
from scipy.optimize import minimize

# define funcao a ser minimizada (recebe um vetor v com x e y)
def funcao(v):
    x, y = v
    return (x - 2)**2 + (y - 3)**2

# encontra o min passando o vetor inicial [0, 0]
resultado = minimize(funcao, [0, 0])

print('minimo encontrado em [x, y] =', resultado.x)
```

**16. Resolva a equação diferencial de segunda ordem d²y/dx² + y = 0, com condições iniciais y(0) = 1 e y′(0) = 0**

```python
from scipy.integrate import solve_ivp
import numpy as np

def sistema_edo(t, y):
    y1, y2 = y
    return [y2, -y1]

# condicoes iniciais y(0) = 1 e y'(0) = 0
condicoes_iniciais = [1, 0]

# Resolver no intervalo de 0 a 10
solucao = solve_ivp(sistema_edo, [0, 10], condicoes_iniciais, t_eval=np.linspace(0, 10, 100))

print('tempo (t) primeiros pontos:\n', solucao.t[:5])
print('\nsolucao para y(t) primeiros pontos:\n', solucao.y[0][:5])
```

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ['A', 'B', 'C', 'D', 'E', 'F']
valores = [15, -10, 20, -5, 10, -20]

# cria lista de cores, sendo verde para positivo e vermelho negativo
cores = ['green' if v > 0 else 'red' for v in valores]

# plota as barras
plt.bar(categorias, valores, color=cores)
plt.axhline(0, color='black', linewidth=1)  # adiciona uma linha preta no 0 para separar
plt.title('Valores positivos e negativos')
plt.show()
```

**19. Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las**

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# cria duas series com parametro label
plt.plot(x, y1, label='Serie 1 Seno')
plt.plot(x, y2, label='Serie 2 Cosseno')
plt.title('Grafico com duas series')
plt.xlabel('eixo X')
plt.ylabel('eixo Y')
plt.legend()
plt.show()
```

**20. Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x²**

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = x**2
# cria o graficp
plt.plot(x, y, color='black', linewidth=2)
plt.title('Grafico de y = x² com grade personalizada')
plt.xticks(np.arange(0, 11, 1))  # marcacoes de 1 em 1 no eixo x
plt.yticks(np.arange(0, 101, 10))  # marcacoes de 10 em 10 no eixo y
plt.grid(color='gray', linestyle=':', linewidth=1)
plt.show()
```
